```
import os
from google import genai

client = genai.Client(
    api_key=os.environ.get("GEMINI_API_KEY"),
)

tools = [
    {
        'type': 'code_execution',
    },
    {
        'type': 'google_search',
    },
    {
        'type': 'url_context',
    },
]

interaction = client.interactions.create(
    agent='antigravity-preview-05-2026',
    input='',
    tools=tools,
    environment={
        'type': 'remote',
        'network': 'disabled',
    },
)

print(interaction.steps[-1])


```
